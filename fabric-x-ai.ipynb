````python
# Import libraries
import re
import json
import requests
from bs4 import BeautifulSoup
````

````python
# Parse website
url = 'https://learn.microsoft.com/en-us/azure/ai-services/agents/concepts/model-region-support'

# Send a GET request to the website
response = requests.get(url)
print(f"response >>>>>{response}")

# Check if the request was successful
if response.status_code == 200:
    # Parse the HTML content of the page
    soup = BeautifulSoup(response.content, 'html.parser')
    body = soup.get_text()
    print(f"body >>>>>{body}")
else:
    print(f"Failed to retrieve the webpage. Status code: {response.status_code}")
````

````python
# This code uses AI. Always review output for mistakes. 
# Read terms: https://azure.microsoft.com/en-us/support/legal/preview-supplemental-terms/

import synapse.ml.aifunc as aifunc
import pandas as pd
import openai

df = pd.DataFrame([
    (body),
], columns=["models_text"])

df["models_json"] = df.ai.generate_response("Extract the 'Azure OpenAI models' table from the input data. Fill in the missing data on each 'region' and set it to false. The output should be a valid JSON using the common column names. Do not extra response that invalidate the JSON.")
display(df)
````

````python
# Extract JSON from df
df_data = df.to_json(orient="records", lines=False)
json_data = json.loads(df_data)
models_json = json_data[0]['models_json']

clean_text = ""

for line in models_json.split("\n"):
    if "```" not in line:
        clean_text += line

clean_json = json.loads(clean_text)
models = clean_json["Azure_OpenAI_Models"]
print(type(models))
````

````python
# Specify the file name (and path if needed)
file_path = f'/lakehouse/default/Files/azure_openai_models.json'

# Save the JSON data
with open(file_path, 'w') as file:
    json.dump(models, file, indent=4)

print(f"Data successfully saved to {file_path}")
````

````python
# Function to clean column names
def clean_column_name(name):
    return re.sub(r'\W+', '-', name.strip().replace(" ", "_"))

# Rename columns
for col_name in df.columns:
    df = df.withColumnRenamed(col_name, clean_column_name(col_name))

# Show the DataFrame with renamed columns
df.show()
````
